```python
# LIVE DEMO — Exploring the Response object
import requests

response = requests.get("https://jsonplaceholder.typicode.com/users/1")

# 1. Status code — did it work?
print("Status code:", response.status_code)  # 200 = OK
print()

# 2. The JSON data — as a Python dict!
data = response.json()
print("Type:", type(data))    # <class 'dict'>
print("Name:", data["name"])  # Leanne Graham
print("Email:", data["email"])
print("City:", data["address"]["city"])  # nested!
print()

# 3. Raw text — for debugging
print("Raw text (first 100 chars):", response.text[:100])
```

```python
# LIVE DEMO — Fetching a list of items
import requests

response = requests.get("https://jsonplaceholder.typicode.com/users")
users = response.json()

print(f"Got {len(users)} users")
print(f"Type: {type(users)}")     # list!
print(f"First user: {type(users[0])}")  # dict!
print()

# It's just a list of dicts — use everything you know
for user in users:
    print(f"  {user['name']} — {user['email']}")
```

```python
# LIVE DEMO — List comprehensions work perfectly with API data
import requests

response = requests.get("https://jsonplaceholder.typicode.com/users")
users = response.json()

# Extract just the names
names = [user["name"] for user in users]
print("Names:", names[:5])

# Extract cities
cities = [user["address"]["city"] for user in users]
print("Cities:", cities[:5])

# Filter — find users whose company catchphrase contains a specific word
tech_users = [user["name"] for user in users if "technology" in user["company"]["bs"]]
print("Tech-related users:", tech_users)
```

```python
# Exercise 1: Fetch and explore JSONPlaceholder posts
import requests

BASE_URL = "https://jsonplaceholder.typicode.com"

# 1. Fetch all posts and print how many there are.
response = requests.get(f"{BASE_URL}/posts", timeout=10)
response.raise_for_status()
posts = response.json()

print(f"1. Total posts: {len(posts)}")

# 2. Print the title of the first and last posts.
print(f"\n2. First post title: {posts[0]['title']}")
print(f"   Last post title: {posts[-1]['title']}")

# 3. Collect titles containing "qui" using a list comprehension.
qui_titles = [
    post["title"]
    for post in posts
    if "qui" in post["title"].lower()
]

print(f"\n3. Number of titles containing 'qui': {len(qui_titles)}")
print("   First 3 matching titles:")
for title in qui_titles[:3]:
    print(f"   - {title}")

# 4. Count the number of posts written by each userId.
posts_per_user = {}
for post in posts:
    user_id = post["userId"]
    posts_per_user[user_id] = posts_per_user.get(user_id, 0) + 1

print("\n4. Posts written by each userId:")
for user_id, post_count in sorted(posts_per_user.items()):
    print(f"   User {user_id}: {post_count} posts")

# 5. Bonus: fetch user 1 and print their name and post count.
user_response = requests.get(f"{BASE_URL}/users/1", timeout=10)
# user_response.raise_for_status()
user_1 = user_response.json()

print(
    f"\n5. {user_1['name']} wrote "
    f"{posts_per_user.get(user_1['id'], 0)} posts."
)
```

```python
# LIVE DEMO — Query parameters: two ways to do it
import requests

# Way 1: Manually in the URL (works, but messy)
response1 = requests.get("https://jsonplaceholder.typicode.com/posts?userId=1")

# Way 2: Using the params dict (cleaner, recommended)
response2 = requests.get(
    "https://jsonplaceholder.typicode.com/posts",
    params={"userId": 1}
)

# Both give the same result
posts_user1 = response2.json()
print(f"Posts by user 1: {len(posts_user1)}")
print(f"First title: {posts_user1[0]['title']}")
print()

# You can see the actual URL that was sent
print(f"URL that was sent: {response2.url}")
```
