```python
import pandas as pd
from plotnine import ggplot, aes, geom_point, labs, theme_minimal

# Sample data
data = pd.DataFrame({
    'x': [1, 2, 3, 4, 5],
    'y': [3, 5, 2, 6, 4]
})

# Create the scatter plot
plot = (
    ggplot(data=data, mapping=aes(x='x', y='y'))
    + geom_point(size=3, color='steelblue')
    + labs(title='Scatter Plot', x='X', y='Y')
    + theme_minimal()
)

# Display inside the notebook
plot
```

```python
from plotnine import geom_smooth

# Adding a smooth line layer
(ggplot(data, aes(x='x', y='y')) + 
 geom_point() + 
 geom_smooth(method='lm'))
```

```python
import pandas as pd
from plotnine import (
    ggplot,
    aes,
    geom_point,
    scale_color_manual,
    labs,
    theme_minimal
)

data = pd.DataFrame({
    'x': [1, 2, 3, 4, 5],
    'y': [3, 5, 2, 6, 4],
    'group': ['A', 'B', 'C', 'A', 'B']
})

plot = (
    ggplot(data, aes(x='x', y='y', color='group'))
    + geom_point(size=4)
    + scale_color_manual(
        values={
            'A': 'blue',
            'B': 'green',
            'C': 'red'
        }
    )
    + labs(title='Scatter Plot by Group', color='Group')
    + theme_minimal()
)

plot
```

```python
from plotnine import theme_bw

(ggplot(data, aes(x='x', y='y')) + 
 geom_point() + 
 theme_bw())
```

```python
import pandas as pd
from plotnine import ggplot, aes, geom_point, facet_wrap, labs, theme_bw

data = pd.DataFrame({
    'x': [1, 2, 3, 4, 5, 6],
    'y': [3, 5, 2, 6, 4, 7],
    'category': ['A', 'A', 'A', 'B', 'B', 'B']
})

plot = (
    ggplot(data, aes(x='x', y='y'))
    + geom_point(size=4, color='steelblue')
    + facet_wrap('~category')
    + theme_bw()
    + labs(
        title='Scatter Plot by Category',
        x='X',
        y='Y'
    )
)

plot
```

```python
import pandas as pd
from plotnine import ggplot, aes, geom_histogram, labs, theme_bw

data = pd.DataFrame({
    'x': [1, 2, 2, 3, 3, 3, 4, 4, 5, 5, 5, 6]
})

plot = (
    ggplot(data, aes(x='x'))
    + geom_histogram(
        binwidth=1,
        boundary=0.5,
        fill='steelblue',
        color='black'
    )
    + theme_bw()
    + labs(
        title='Distribution of X',
        x='X',
        y='Frequency'
    )
)

plot
```

```python
import plotly.express as px
import pandas as pd

# Sample data
data = pd.DataFrame({'x': [1, 2, 3, 4], 'y': [10, 11, 12, 13]})

# Creating a line chart
fig = px.line(data, x='x', y='y')
fig.show()
```

```python
# Creating an interactive scatter plot
fig = px.scatter(data, x='x', y='y', color='y')
fig.show()
```

```python
fig.update_layout(title='Customized Plot', 
                  xaxis_title='X Axis', 
                  yaxis_title='Y Axis',
                  template='plotly_dark')
fig.show()
```

```python
from plotly.subplots import make_subplots
import plotly.graph_objects as go

# Creating subplots
fig = make_subplots(rows=1, cols=2)
fig.add_trace(go.Scatter(x=data['x'], y=data['y']), row=1, col=1)
fig.add_trace(go.Bar(x=data['x'], y=data['y']), row=1, col=2)
fig.show()
```

```python
from dash import Dash, dcc, html

# Create the Dash application using the Plotly figure from the previous cell.
app = Dash(__name__)

app.layout = html.Div([
    html.H3('Interactive Plotly Dashboard'),
    dcc.Graph(figure=fig)
])

# Display the dashboard directly inside Jupyter Notebook.

app.run(jupyter_mode='inline', debug=False)
```

```python
import pandas as pd
import plotly.express as px

data_3d = pd.DataFrame({
    'x': [1, 2, 3, 4],
    'y': [10, 11, 12, 13],
    'z': [5, 8, 3, 7]
})

fig = px.scatter_3d(
    data_3d,
    x='x',
    y='y',
    z='z',
    color='y',
    title='Interactive 3D Scatter Plot'
)

fig.update_traces(marker=dict(size=7))
fig.show()
```

```python
import numpy as np
import pandas as pd
from bokeh.io import output_notebook, show
from bokeh.plotting import figure

# Create a reproducible larger dataset and sample it for visualization.
rng = np.random.default_rng(seed=42)
large_df = pd.DataFrame({
    'x_column': rng.normal(size=5000),
    'y_column': rng.normal(size=5000)
})
sampled_df = large_df.sample(n=500, random_state=42)

# Render Bokeh output inside the notebook.
output_notebook()

p = figure(
    title='Sampled Dataset Visualization with Bokeh',
    width=800,
    height=500,
    x_axis_label='X',
    y_axis_label='Y',
    tools='pan,wheel_zoom,box_zoom,reset,save'
)

p.scatter(
    x=sampled_df['x_column'],
    y=sampled_df['y_column'],
    marker='circle',
    size=8,
    color='steelblue',
    alpha=0.6
)

show(p)
```
