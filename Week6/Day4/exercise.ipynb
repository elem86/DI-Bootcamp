🌟 Exercise 1 : Matrix Operations

Instructions

In this exercise, you’ll work with a 3x3 matrix. Here’s a brief explanation of the concepts:

Determinant: The determinant is a value that can be computed from the elements of a square matrix. It provides important information about the matrix, such as whether it has an inverse, and is used in various areas like linear algebra and calculus. To understand more about it you can watch this video.

Inverse of a Matrix: The inverse of a matrix is a matrix that, when multiplied with the original matrix, results in an identity matrix. Not all matrices have inverses. The inverse is crucial in solving systems of linear equations.
Create a 3x3 matrix and perform the following operations:

 - Calculate the determinant.
 - Find the inverse of the matrix.

```python
#Calculate the determinant.
import numpy as np
import pandas as pd
# Define a 3x3 matrix
matrix = np.array([[1, 2, 3], 
                   [0, 1, 4], 
                   [5, 6, 0]])

# Calculate the determinant
determinant = np.linalg.det(matrix)

print(f"The determinant is: {determinant}")
```

```python
#Find the inverse of the matrix.
matrix = np.array([[1, 2, 3], 
                   [0, 1, 4], 
                   [5, 6, 0]])
try:
    # Calculate the inverse
    inverse_matrix = np.linalg.inv(matrix)
    print("The inverse matrix is:")
    print(inverse_matrix)
    
    # Verification: Matrix * Inverse should equal the Identity Matrix
    identity = np.dot(matrix, inverse_matrix)
    print("\nVerification (should be close to Identity Matrix):")
    print(identity)

except np.linalg.LinAlgError:
    print("This matrix is singular and cannot be inverted.")
```

🌟 Exercise 5 : Image Representation

Instructions

Explain how images are represented in NumPy arrays and demonstrate with a simple example (e.g., creating a 5x5 grayscale image).

How Images are Represented
 - Grayscale Images: Represented as a 2D array with shape (height, width).
   - Each element is a single integer (usually uint8) ranging from 0 (black) to 255 (white).
   - Example: A pixel value of 128 is a medium gray.
 - Color (RGB) Images: Represented as a 3D array with shape (height, width, 3).
   - The third dimension represents the Red, Green, and Blue channels.
   - Each pixel is a vector like [R, G, B], e.g., [255, 0, 0] is pure red. 

```python
import matplotlib.pyplot as plt

# 1. Create a 5x5 NumPy array with specific pixel values (0-255)
# Let's create a simple pattern: a white border with a black center
image = np.ones((5, 5), dtype=np.uint8) * 255  # Start with all white (255)
image[1:4, 1:4] = 0                            # Set the center 3x3 to black (0)

# Alternatively, create a random grayscale image:
# image = np.random.randint(0, 256, size=(5, 5), dtype=np.uint8)

print("Image Array Shape:", image.shape)
print("Image Data:\n", image)

# 2. Display the array as an image
plt.imshow(image, cmap='gray') # 'cmap='gray'' ensures it renders as grayscale
plt.title("5x5 Grayscale Image")
plt.axis('off') # Hide axis numbers for cleaner look
plt.show()   
```

🌟 Exercise 6 : Basic Hypothesis Testing
Instructions

Create a sample dataset to test the effectiveness of a new training program on employee productivity:


import numpy as np

#Productivity scores of employees before the training program
productivity_before = np.random.normal(loc=50, scale=10, size=30)

#Productivity scores of the same employees after the training program
productivity_after = productivity_before + np.random.normal(loc=5, scale=3, size=30)

#Your task is to formulate a hypothesis regarding the training program's effectiveness 
#and test it using basic statistical functions in NumPy.


Given a dataset, formulate a simple hypothesis and test it using basic statistical functions in NumPy.

```python
# Productivity scores of employees before the training program
productivity_before = np.random.normal(loc=50, scale=10, size=30)

# Productivity scores of the same employees after the training program
productivity_after = productivity_before + np.random.normal(loc=5, scale=3, size=30)

# Your task is to formulate a hypothesis regarding the training program's effectiveness 
# and test it using basic statistical functions in NumPy.
```

H0: The training program did not help with the productivity scores.
H1: The training program helped with the productivity scores.

P value = 5%

```python
from scipy import stats
# Perform Paired t-test
t_stat, p_value = stats.ttest_rel(productivity_before, productivity_after)

print(f"T-statistic: {t_stat:.2f}")
print(f"P-value: {p_value:.2f}")

if p_value < 0.05:
    print("Result: Statistically significant difference between the before and after training performance (reject null hypothesis).")
else:
    print("Result: No statistically significant difference.")
```

Demonstrate how to convert a NumPy array to a Pandas DataFrame and vice versa.

```python
#Convert NumPy Array to Pandas DataFrame

# Create a 2D NumPy array
data = np.array([[10, 20, 30], 
                 [40, 50, 60], 
                 [70, 80, 90]])

# Convert to DataFrame with custom column names
df = pd.DataFrame(data, columns=['A', 'B', 'C'])

print("Original NumPy Array:")
print(data)
print("\nConverted Pandas DataFrame:")
print(df)
```
